```python
volume = [35, 42, 48, 45]
total = sum(volume)
rata_rata = sum(volume) / len(volume)
tertinggi = max(volume)
terendah = min(volume)

print("Total:", total)
print("Rata-rata:", rata_rata)
print("Tertinggi:", tertinggi)
print("Terendah:", terendah)
```

```python
import numpy as bdmd

volume = bdmd.array([35, 42, 48, 45])
print("Total:", bdmd.sum(volume))
print("Rata-rata:", bdmd.mean(volume))
print("Paling Tinggi:", bdmd.max(volume))
print("Paling Rendah:", bdmd.min(volume))
```

```python
import numpy as np

panjang = np.array([4.0, 5.0, 6.0])  #meter
lebar = np.array([3.0, 3.0, 2.5])    #meter
tebal = 0.12                           #meter

luas = panjang * lebar
volume = luas * tebal

print("Luas tiap pelat:", luas, "m2")
print("Volume tiap pelat:", volume, "m3")
print("Total volume:", np.sum(volume), "m3")
```
